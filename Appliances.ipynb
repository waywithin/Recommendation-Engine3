## Popularity Based Recommendation System

```python
avg_ratings = products.groupby("name")["ratings"].mean().reset_index()
avg_ratings.rename(columns={"ratings":"avg_ratings"}, inplace=True)
```

```python
combine = avg_ratings.merge(products, on = "name").drop_duplicates()[["name", "main_category", "sub_category", "image", "link", "no_of_ratings", "avg_ratings"]]
```

```python
combine
```

```python
popular_appliances = combine[combine["no_of_ratings"]>250].sort_values("avg_ratings", ascending = False).head(50)
popular_appliances.head()
```

## **Content Based Filtering**

```python
products.head(2)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import linear_kernel
```

```python
tfidf = TfidfVectorizer(stop_words="english")
```

```python
products["tags"] = (products["name"] + " " + products["main_category"] + " " + products["sub_category"])
```

```python
products.head()
```

```python
tfidf_matrix = tfidf.fit_transform(products["tags"])
```

```python
def recommend(user_input):
    query_vector = tfidf.transform([user_input])
    cosine_sim = linear_kernel(tfidf_matrix, query_vector)
    sim_score = sorted(list(enumerate(cosine_sim)), key = lambda x:x[1], reverse = True)[1:10]
    sim_index = [i[0] for i in sim_score]

    print(products["tags"].iloc[sim_index])

recommend("appliances")
```
